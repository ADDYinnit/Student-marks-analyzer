## Student Marks Dataset - Exploratory Data Analysis (EDA)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")

# Load the dataset
df = pd.read_csv('/content/Student_Marks.csv')

# Display basic information and first few rows
print("--- Dataset Info ---")
df.info()

print("\n--- Missing Values ---")
print(df.isnull().sum())

print("\n--- First 5 Rows ---")
display(df.head())
```

```python
# Descriptive Statistics
print("--- Summary Statistics ---")
display(df.describe())
```

### Visualizing Distributions and Relationships

```python
# Check if the expected columns exist before plotting
columns = df.columns.tolist()
print(f"Columns available: {columns}")

# Automatically determine numerical columns to plot
num_cols = df.select_dtypes(include=['number']).columns.tolist()

# Pairplot to see pairwise relationships and distributions
if len(num_cols) > 1:
    sns.pairplot(df)
    plt.suptitle("Pairwise Relationships & Distributions", y=1.02)
    plt.show()
```

```python
# Correlation heatmap for numerical features
if len(num_cols) > 1:
    plt.figure(figsize=(8, 6))
    sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
    plt.title('Correlation Matrix')
    plt.show()
```
